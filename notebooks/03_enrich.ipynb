# 03 Enrichment

This step enriches hosts with passive fingerprints and optional metadata mapping.

## What happens here
- collect available fingerprint signals (JA3/JA3S/HASSH/SNI)
- optionally map fingerprints to CPE identifiers
- optionally use p0f for passive OS hints
- save enriched host records

## Input and output
- input: `run/normalized/flows.jsonl` (+ optional PCAP/CPE map)
- output: `run/enriched/enriched_hosts.jsonl`

## How to use the panel
- `Run Enrich`: execute enrichment
- `Refresh Preview`: display fingerprint summaries and top values
- `Preview N`: changes only the number of displayed rows


### Environment initialization
Prepares paths and imports for enrichment.


```python
from pathlib import Path
import os
import sys


def find_project_root(start: Path) -> Path:
    candidates = [start, *start.parents]

    env_root = os.environ.get("PMMAP_ROOT")
    if env_root:
        candidates.append(Path(env_root))

    candidates.append(Path.home() / "Bakalarka" / "passive-network-mapping-platform")

    seen = set()
    for path in candidates:
        try:
            path = path.resolve()
        except Exception:
            pass
        key = str(path)
        if key in seen:
            continue
        seen.add(key)
        if (path / "pmmap").is_dir():
            return path

    raise RuntimeError(
        "Could not locate project root (directory containing 'pmmap'). "
        "Open notebook from repository workspace or set PMMAP_ROOT."
    )


ROOT = find_project_root(Path().resolve())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from pmmap.notebook_common import init_notebook_paths

ROOT, DATA_DIR, RUN_DIR = init_notebook_paths(start=ROOT)
```
