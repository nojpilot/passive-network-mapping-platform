# 02 Host Inventory

This step aggregates normalized flows into host-level inventory records.

## What happens here
- group flow records by host IP
- compute traffic volume and flow counts per host
- infer basic host role hints (for example DNS/NTP/DHCP behavior)

## Input and output
- input: `run/normalized/flows.jsonl`
- output: `run/inventory/hosts.jsonl`

## How to use the panel
- `Run Inventory`: build host inventory
- `Refresh Preview`: display top hosts and role distribution
- `Preview N`: changes only preview tables/charts


### Environment initialization
Prepares paths and imports for the inventory step.


```python
from pathlib import Path
import os
import sys


def find_project_root(start: Path) -> Path:
    candidates = [start, *start.parents]

    env_root = os.environ.get("PMMAP_ROOT")
    if env_root:
        candidates.append(Path(env_root))

    candidates.append(Path.home() / "Bakalarka" / "passive-network-mapping-platform")

    seen = set()
    for path in candidates:
        try:
            path = path.resolve()
        except Exception:
            pass
        key = str(path)
        if key in seen:
            continue
        seen.add(key)
        if (path / "pmmap").is_dir():
            return path

    raise RuntimeError(
        "Could not locate project root (directory containing 'pmmap'). "
        "Open notebook from repository workspace or set PMMAP_ROOT."
    )


ROOT = find_project_root(Path().resolve())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from pmmap.notebook_common import init_notebook_paths

ROOT, DATA_DIR, RUN_DIR = init_notebook_paths(start=ROOT)
```
